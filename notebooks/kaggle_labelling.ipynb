#### Full LLM Labelling Run (Kaggle launcher)

- runs `scripts/run_labelling.py` from the repo on Kaggle's GPUs. the labelling code lives in `src/quorum/labelling.py` and `src/quorum/evaluation.py`.
- needs: GPU T4 x2, internet on, the pool dataset added as input, and the `HF_TOKEN` secret attached.

```python
import os
from kaggle_secrets import UserSecretsClient

os.environ["HF_TOKEN"] = UserSecretsClient().get_secret("HF_TOKEN")      # used by hugging face for gated models
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"     # reduces GPU memory fragmentation
```

```python
# label the full pool with all 3 models (about an hour). -u prints progress as it happens

!cd quorum && PYTHONPATH=src python -u scripts/run_labelling.py \
    --pool /kaggle/input/datasets/thisisbeancurd/quorum-pool-test-ds/pool.csv \
    --out-dir /kaggle/working
```
